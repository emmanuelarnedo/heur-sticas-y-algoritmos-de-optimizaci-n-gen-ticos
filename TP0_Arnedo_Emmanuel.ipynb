#### **Tarea**: generar un vector con n elementos generados al azar.

```python
import random

def aleatorio_1(n):
    """Método 1: dinámico usando bucle for
    """
    v = []
    for i in range(n):
      v.append(random.random())
    return v

def aleatorio_2(n):
    """Método 2:  inicializando e iterando
    """
    v = [0]*n
    for i in range(len(v)):
      v[i] = random.random()
    return v

def aleatorio_3(n):
    """Método 2: list comprehensions
    """
    return [random.random() for i in range(n)]

n = 6
print('Método 1 (con for dinámico)')
print(aleatorio_1(n))
print('Método 2 (con inicializacion)')
print(aleatorio_2(n))
print('Método 3 (list comprehensions)')
print(aleatorio_3(n))
```

1.    Implementar un programa que iterativamente genere numeros aleatorios enteros dentro del intervalo [0, 20] y que termine la ejecución al obtener un 0. Debe imprimir en pantalla la cantidad de iteraciones realizadas al terminar la ejecución.




```python
import random

def contar_hasta_cero():
    contador_iteraciones = 0
    numero = -1  # Inicializamos con un valor diferente de 0 para entrar al bucle

    print("Generando números aleatorios entre 0 y 20...")

    while numero != 0:
        numero = random.randint(0, 20)
        contador_iteraciones += 1
        print(f"Número obtenido: {numero}")

    print(f"\n¡Se obtuvo un 0! El programa terminó después de {contador_iteraciones} iteraciones.")

# Ejecutar la función
contar_hasta_cero()
```

2. Generar un vector con n elementos binarios (valor 0 o 1 generado al azar). Luego contar cuantos ceros hay en la primera mitad del vector (posición 0 a n/2) y cuantos unos hay en la segunda mitad del vector. Probar con n = 10 y n = 13

```python
import random

def analizar_vector_binario(n):
    # Generar vector con n elementos binarios (0 o 1)
    vector = [random.randint(0, 1) for _ in range(n)]

    # Calcular el punto medio (división entera para manejar n impar)
    mitad = n // 2

    # Primera mitad (de 0 a n/2 - 1) y segunda mitad (de n/2 a n - 1)
    primera_mitad = vector[:mitad]
    segunda_mitad = vector[mitad:]

    # Contar ceros en la primera mitad y unos en la segunda mitad
    ceros_primera = primera_mitad.count(0)
    unos_segunda = segunda_mitad.count(1)

    # Mostrar resultados
    print(f"Prueba con n = {n}")
    print(f"Vector generado: {vector}")
    print(f"Primera mitad:   {primera_mitad}")
    print(f"Segunda mitad:   {segunda_mitad}")
    print(f"Ceros en la primera mitad: {ceros_primera}")
    print(f"Unos en la segunda mitad:  {unos_segunda}\n")

# Probar con n = 10 y n = 13
analizar_vector_binario(10)
analizar_vector_binario(13)
```

## Ejercicio 2
Graficar la norma del vector para una corrida con 10, 50, 100, y 500 puntos aleatorios. De ser necesario, buscar lo que se denomina "norma de un vector".

```python
import math
import random
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
from typing import List, Tuple

# %matplotlib inline

def generar_vectores_2d(cantidad: int, limite: int = 100) -> List[Tuple[float, float]]:
    # Genera coordenadas (x, y) aleatorias uniformemente distribuidas.
    return [(random.uniform(-limite, limite), random.uniform(-limite, limite))
            for _ in range(cantidad)]

def graficar_analisis_completo(corridas: List[int]) -> None:
    """
    Genera una figura con 2 columnas por cada corrida:
    Columna 1: Representación en el plano cartesiano 2D.
    Columna 2: Histograma de distribución de la norma euclidiana.
    """
    # Creamos una cuadrícula dinámica según la cantidad de corridas solicitadas
    fig, axs = plt.subplots(len(corridas), 2, figsize=(14, 4 * len(corridas)))
    fig.suptitle("Análisis de Vectores Aleatorios: Plano Físico vs. Distribución de su Norma", fontsize=16)

    for i, cantidad in enumerate(corridas):
        vectores = generar_vectores_2d(cantidad)

        normas = [math.hypot(x, y) for x, y in vectores]

        ax_2d = axs[i, 0]
        xs = [v[0] for v in vectores]
        ys = [v[1] for v in vectores]

        segmentos = [[(0, 0), (x, y)] for x, y in vectores]
        coleccion = LineCollection(segmentos, colors='blue', linewidths=0.6, alpha=0.5)
        ax_2d.add_collection(coleccion)

        ax_2d.scatter(xs, ys, color='red', s=10, zorder=3)

        ax_2d.axhline(0, color='black', linewidth=1)
        ax_2d.axvline(0, color='black', linewidth=1)
        ax_2d.set_xlim(-110, 110)
        ax_2d.set_ylim(-110, 110)
        ax_2d.set_aspect('equal')
        ax_2d.set_title(f"Representación Espacial (N={cantidad})")
        ax_2d.set_xlabel("Coordenada X")
        ax_2d.set_ylabel("Coordenada Y")

        ax_hist = axs[i, 1]
        ax_hist.hist(normas, bins=15, color='steelblue', edgecolor='black', alpha=0.8)

        ax_hist.set_title(f"Distribución de la Norma ||v|| (N={cantidad})")
        ax_hist.set_xlabel("Magnitud de la Norma")
        ax_hist.set_ylabel("Frecuencia (Cantidad de vectores)")
        ax_hist.grid(axis='y', linestyle=':', alpha=0.6)

    plt.tight_layout()
    plt.show()

tamanos_corridas = [10, 50, 100, 500]
graficar_analisis_completo(tamanos_corridas)
```
